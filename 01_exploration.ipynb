```python
import pybaseball
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
lookup = pybaseball.playerid_lookup('Yamamoto', 'Yoshinobu')
print("Lookup result:")
print(lookup)

if lookup.empty:
    raise ValueError("playerid_lookup returned no rows. Check name/order or try alternate lookup.")

mlbam_id = lookup['key_mlbam'].iloc[0]
print("\nMLBAM ID:", mlbam_id)

start_date = '2024-03-30'
end_date = '2024-10-01'
Yamamoto_data_2024 = pybaseball.statcast_pitcher(start_date, end_date, player_id=int(mlbam_id))

print("\nColumns:")
print(list(Yamamoto_data_2024.columns))

print("\nFirst 5 rows:")
print(Yamamoto_data_2024.head(5))
```

```python
lookup = pybaseball.playerid_lookup('Stone', 'Gavin')
print("Lookup result:")
print(lookup)

if lookup.empty:
    raise ValueError("playerid_lookup returned no rows. Check name/order or try alternate lookup.")

mlbam_id = lookup['key_mlbam'].iloc[0]
print("\nMLBAM ID:", mlbam_id)

start_date = '2024-03-30'
end_date = '2024-10-01'
Stone_data_2024 = pybaseball.statcast_pitcher(start_date, end_date, player_id=int(mlbam_id))

print("\nColumns:")
print(list(Stone_data_2024.columns))

print("\nFirst 5 rows:")
print(Stone_data_2024.head(5))
```

```python
lookup = pybaseball.playerid_lookup('Glasnow', 'Tyler')
print("Lookup result:")
print(lookup)

if lookup.empty:
    raise ValueError("playerid_lookup returned no rows. Check name/order or try alternate lookup.")

mlbam_id = lookup['key_mlbam'].iloc[0]
print("\nMLBAM ID:", mlbam_id)

start_date = '2024-03-30'
end_date = '2024-10-01'
Glasnow_data_2024 = pybaseball.statcast_pitcher(start_date, end_date, player_id=int(mlbam_id))

print("\nColumns:")
print(list(Glasnow_data_2024.columns))

print("\nFirst 5 rows:")
print(Glasnow_data_2024.head(5))
```

```python
results = pd.concat([Yamamoto_data_2024, Stone_data_2024, Glasnow_data_2024], ignore_index=True)
print("Rows in combined dataset:", len(results))
print ("Rows in Yamamoto dataset:", len(Yamamoto_data_2024))
print ("Rows in Stone dataset:", len(Stone_data_2024))
print ("Rows in Glasnow dataset:", len(Glasnow_data_2024))
results.dropna(subset=['pitch_type'], inplace=True)
results_filtered = results[results['pitch_type'] != 'ST']
print("Pitch Frequency:", results_filtered['pitch_type'].value_counts())
```

```python
physical_attributes = ['release_speed', 'release_pos_x', 'release_pos_y', 'release_pos_z', 'arm_angle', 'effective_speed', 'release_spin_rate', 'spin_axis', 'release_extension', 'pfx_x', 'pfx_z']
outcomes = ['events', 'description', 'type', 'launch_speed', 'launch_angle', 'launch_direction', 'hit_distance_sc', 'hit_location', 'estimated_ba_using_speedangle', 'estimated_woba_using_speedangle', 'woba_value', 'babip_value']
```

```python
sns.kdeplot(data=results_filtered, x='release_speed', hue='pitch_type', fill=True)
plt.title('Release Speed Distribution by Pitch Type')
plt.xlabel('Release Speed (mph)')
plt.ylabel('Density')
```
