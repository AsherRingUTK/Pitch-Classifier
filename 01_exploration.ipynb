```python
import pybaseball
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# lookup a pitcher (example: Gerrit Cole), get his MLBAM id, pull 2024 pitch-level data, and inspect it
lookup = pybaseball.playerid_lookup('Ohtani', 'Shohei')
print("Lookup result:")
print(lookup)

if lookup.empty:
    raise ValueError("playerid_lookup returned no rows. Check name/order or try alternate lookup.")

mlbam_id = lookup['key_mlbam'].iloc[0]
print("\nMLBAM ID:", mlbam_id)

start_date = '2025-03-30'
end_date = '2025-10-01'
pitcher_data_2025 = pybaseball.statcast_pitcher(start_date, end_date, player_id=int(mlbam_id))

print("\nColumns:")
print(list(pitcher_data_2025.columns))

print("\nFirst 5 rows:")
print(pitcher_data_2025.head(5))
```
